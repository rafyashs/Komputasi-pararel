```python
n =  5
total = 0
print("Serial Computation")
for i in range (1, n+1):
	total += i
	print(f"step {i}: total = {total} ")
print("Final Serial sum is" ,total)
```

```python
import multiprocessing

def hitung_sum(x, y):
    return x + y

if __name__ == "__main__":
    # Data input
    a, b, c, d = 10, 20, 30, 40

    # Membuat pool proses (menggunakan 2 core)
    with multiprocessing.Pool(processes=2) as pool:
        # Menjalankan Sum1 dan Sum2 secara paralel
        # pool.apply_async akan mengirim tugas ke core yang tersedia
        res1 = pool.apply_async(hitung_sum, (a, b))
        res2 = pool.apply_async(hitung_sum, (c, d))

        # Mengambil hasil
        sum1 = res1.get()
        sum2 = res2.get()

    # Langkah terakhir: Final Sum
    final_sum = sum1 + sum2

    print(f"Sum1: {sum1}")
    print(f"Sum2: {sum2}")
    print(f"Final Sum: {final_sum}")
```
