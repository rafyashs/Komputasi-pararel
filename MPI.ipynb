```python
%%writefile latihan_mpi.py
from mpi4py import MPI

# Inisialisasi komunikator global
comm = MPI.COMM_WORLD
rank = comm.Get_rank()  # ID unik setiap proses
size = comm.Get_size()  # Total jumlah proses yang berjalan

if rank == 0:
    # Proses Master (Rank 0) menyiapkan data
    data_paket = {'id': 1, 'konten': 'Informasi dari Master'}
    print(f"--- Master (Rank {rank}) bersiap mengirim data ke Rank 1 ---")
    comm.send(data_paket, dest=1, tag=7)
    print(f"Master (Rank {rank}): Data terkirim.")

elif rank == 1:
    # Proses Worker (Rank 1) menunggu data
    print(f"--- Worker (Rank {rank}) menunggu kiriman... ---")
    data_terima = comm.recv(source=0, tag=7)
    print(f"Worker (Rank {rank}) berhasil menerima: {data_terima}")

else:
    # Rank lain (jika ada) hanya melapor aktif
    print(f"Proses Rank {rank} aktif (tidak ada tugas).")
```
